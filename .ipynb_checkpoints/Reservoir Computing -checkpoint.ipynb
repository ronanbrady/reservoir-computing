# 4、Examples of RC

## 4.1 ESN

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

def spectral_radius(W):
    return np.max(np.abs(np.linalg.eigvals(W)))

class EchoStateNetwork:
    def __init__(
        self,
        n_inputs,
        n_reservoir,
        n_outputs,
        spectral_radius_target=0.95,
        input_scale=0.6,
        leak_rate=0.3,
        ridge_alpha=1e-6,
        sparsity=0.9,
        seed=42,
        use_bias=True,
        include_inputs_in_readout=True
    ):
        self.n_reservoir = n_reservoir
        self.leak_rate = leak_rate
        self.ridge_alpha = ridge_alpha
        self.use_bias = use_bias
        self.include_inputs_in_readout = include_inputs_in_readout

        rng = np.random.default_rng(seed)

        self.W_in = rng.uniform(-1, 1, (n_reservoir, n_inputs)) * input_scale

        W = rng.uniform(-1, 1, (n_reservoir, n_reservoir))
        mask = rng.random(W.shape) < sparsity
        W[mask] = 0.0
        self.W_res = W * (spectral_radius_target / spectral_radius(W))

        self.W_out = None

    def _update(self, x, u):
        return (1 - self.leak_rate) * x + self.leak_rate * np.tanh(
            self.W_res @ x + self.W_in @ u
        )

    def _feature(self, x, u):
        parts = []
        if self.use_bias:
            parts.append(np.array([1.0]))
        parts.append(x)
        if self.include_inputs_in_readout:
            parts.append(u)
        return np.concatenate(parts)

    def fit(self, U, Y, washout=100):
        x = np.zeros(self.n_reservoir)
        X, Yt = [], []

        for t in range(len(U)):
            x = self._update(x, U[t])
            if t >= washout:
                X.append(self._feature(x, U[t]))
                Yt.append(Y[t])

        X = np.vstack(X)
        Yt = np.vstack(Yt)

        self.W_out = np.linalg.solve(
            X.T @ X + self.ridge_alpha * np.eye(X.shape[1]),
            X.T @ Yt
        )

    def predict(self, U):
        x = np.zeros(self.n_reservoir)
        Yhat = []

        for u in U:
            x = self._update(x, u)
            Yhat.append(self._feature(x, u) @ self.W_out)

        return np.vstack(Yhat)
```

## 4.2 Example 1: Sine Wave One-Step Prediction

```python


# ---------- 1. 生成数据 ----------
T = 3000
t = np.arange(T)
signal = np.sin(2*np.pi*t/50) + 0.3*np.sin(2*np.pi*t/10)

U = signal[:-1].reshape(-1, 1)
Y = signal[1:].reshape(-1, 1)

# ---------- 2. 划分训练 / 测试 ----------
split = 2000
U_train, Y_train = U[:split], Y[:split]
U_test, Y_test = U[split:], Y[split:]

# ---------- 3. 训练 ESN ----------
esn = EchoStateNetwork(n_inputs=1, n_reservoir=400, n_outputs=1)
esn.fit(U_train, Y_train)

# ---------- 4. 预测 ----------
Y_pred = esn.predict(U_test)

# ---------- 5. 计算 NMSE ----------
def nmse(y_true, y_pred):
    mse = np.mean((y_true - y_pred) ** 2)
    var = np.var(y_true)
    return mse / var

nmse_value = nmse(Y_test, Y_pred)
print("NMSE:", nmse_value)

# ---------- 6. 画图 ----------
plt.figure()
plt.plot(Y_test[:300], label="True")
plt.plot(Y_pred[:300], label="Predicted")
plt.title(f"Sine Wave One-Step Prediction using ESN (NMSE = {nmse_value:.4f})")
plt.legend()
plt.show()
```

```python

def nmse(y_true, y_pred):
    y_true = np.asarray(y_true).reshape(-1)
    y_pred = np.asarray(y_pred).reshape(-1)
    mse = np.mean((y_true - y_pred)**2)
    var = np.var(y_true)
    return mse / (var + 1e-12)

# --- make sure 1D arrays ---
y_true = np.asarray(Y_test).reshape(-1)
y_pred = np.asarray(Y_pred).reshape(-1)

e = y_true - y_pred
score = nmse(y_true, y_pred)

# --- plot: prediction + residual ---
N = 300  # show first N test points (change if you like)

plt.figure(figsize=(8, 4))
plt.plot(y_true[:N], label="True")
plt.plot(y_pred[:N], label="Predicted")
plt.title(f"Sine Wave One-Step Prediction (ESN)  |  NMSE={score:.4g}")
plt.xlabel("Time step")
plt.ylabel("Amplitude")
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 3))
plt.plot(e[:N], label="Residual (True - Predicted)")
plt.axhline(0.0)  # do not specify color per your environment rules
plt.title("Residual Plot (ESN on Sine Wave)")
plt.xlabel("Time step")
plt.ylabel("Error")
plt.legend()
plt.tight_layout()
plt.show()
```

## 4.3 Example 2: NARMA10 Benchmark

```python

def generate_narma10(T):
    rng = np.random.default_rng(0)
    u = rng.uniform(0.0, 0.5, T)
    y = np.zeros(T)
    for t in range(10, T-1):
        y[t+1] = (
            0.3*y[t]
            + 0.05*y[t]*np.sum(y[t-9:t+1])
            + 1.5*u[t-9]*u[t]
            + 0.1
        )
    return u.reshape(-1,1), y.reshape(-1,1)

# ---------- data ----------
U, Y = generate_narma10(4500)

split = 3000
U_train, Y_train = U[:split], Y[:split]
U_test, Y_test = U[split:], Y[split:]

# ---------- train ----------
esn = EchoStateNetwork(
    n_inputs=1, n_reservoir=800, n_outputs=1,
    spectral_radius_target=0.9, leak_rate=0.2
)
esn.fit(U_train, Y_train, washout=200)

# ---------- predict ----------
Y_pred = esn.predict(U_test)

# ---------- NMSE (recommended: ignore initial test transient) ----------
test_washout = 100
nmse_value = nmse(Y_test[test_washout:], Y_pred[test_washout:])
print("NARMA10 NMSE:", nmse_value)

# ---------- plot (focus on steady region) ----------
start = 100   # skip early transient in the plot
end = 400
plt.figure(figsize=(8, 4))
plt.plot(Y_test[start:end], label="True")
plt.plot(Y_pred[start:end], label="Predicted")
plt.title(f"NARMA10 Prediction using ESN (NMSE = {nmse_value:.3f})")
plt.xlabel("Time step")
plt.ylabel("Output")
plt.legend()
plt.tight_layout()
plt.show()
```
